## 這篇 Notebook 在做什麼

這頁用一個小實驗示範 Jupyter Notebook 的三種基本元素：Markdown 說明、Python 程式碼、執行後的輸出。

資料是隨機產生的三維點。程式碼會算出基本統計量、畫一張散佈圖，最後用 pandas 整理成表格。整頁在本地執行，瀏覽器只負責顯示結果。

## 準備環境

先匯入 NumPy、Matplotlib 與 pandas，並固定隨機種子。固定種子讓每次重建的圖與數字一致，方便對照。

```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(0)

plt.rcParams["font.sans-serif"] = [
    "PingFang TC",
    "Heiti TC",
    "Arial Unicode MS",
    "DejaVu Sans",
]
plt.rcParams["axes.unicode_minus"] = False

print("NumPy", np.__version__)
print("pandas", pd.__version__)
print("Matplotlib", plt.matplotlib.__version__)
```

## 產生資料

用 `np.random.normal` 產生三組常態分布資料，再把三軸組合成一個 `(100, 3)` 的陣列。這一步只用到 NumPy，不需要外部檔案。

```python
n = 100
x = np.random.normal(0, 1, n)
y = np.random.normal(0, 1, n)
z = np.random.normal(0, 1, n)

points = np.column_stack([x, y, z])
points.shape
```

```python
points[:5].round(2)
```

## 畫一張散佈圖

下圖把三軸兩兩配對，畫成三張子圖。顏色代表 `z` 軸數值，方便看出第三維的變化。

```python
fig, axes = plt.subplots(
    1, 3, figsize=(12, 4), dpi=150, constrained_layout=True
)
pairs = [("x", "y", x, y), ("x", "z", x, z), ("y", "z", y, z)]

for ax, (xlabel, ylabel, a, b) in zip(axes, pairs):
    scatter = ax.scatter(
        a,
        b,
        c=z,
        cmap="viridis",
        s=28,
        alpha=0.82,
        edgecolors="white",
        linewidths=0.4,
    )
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    ax.grid(alpha=0.25)

fig.colorbar(scatter, ax=axes, shrink=0.82, label="z")
fig.suptitle("三維隨機資料的三組投影", fontsize=13)
fig.patch.set_facecolor("white")
plt.show()
```

## 整理成表格

最後把三軸的平均、標準差與相關係數放進 DataFrame。Notebook 會把 DataFrame 轉成 HTML 表格，直接顯示在輸出區。

```python
summary = pd.DataFrame(
    {
        "平均": points.mean(axis=0),
        "標準差": points.std(axis=0),
        "最小值": points.min(axis=0),
        "最大值": points.max(axis=0),
    },
    index=["x", "y", "z"],
)

summary.round(2)
```

## 怎麼讀這張圖

三張子圖共用同一個 `z` 色階。顏色越亮，`z` 越大。三個平面各自看得到不同的分布形狀，合起來才是完整的三維資料。

散佈圖適合看群集與離群點，不適合直接讀出精確數值。需要數字時，看上一節的表格。
